# Funciones básicas de Python: funciones 21 a 30

En este notebook se explican diez funciones incorporadas de Python, desde `all()` hasta `open()`. Los ejemplos se relacionan con situaciones de Medicina Veterinaria, como revisar requisitos de alimentación, organizar registros y consultar datos de animales.

## 21. `all()`

**¿Para qué sirve?**  
Devuelve `True` si todos los elementos de un iterable son verdaderos; si alguno es falso, devuelve `False`.

**Sintaxis general:** ``all(iterable)``

**Ejemplo práctico:**

Comprueba si todas las raciones formuladas cumplen el requisito mínimo de proteína definido para este ejercicio.



```python
'proteina_porcentaje = [16.0, 17.5, 16.8]
requisito_minimo = 16.0

cumplen_requisito = all(proteina >= requisito_minimo for proteina in proteina_porcentaje)
print("¿Todas las raciones cumplen el mínimo de proteína?", cumplen_requisito)'

```

**Resultado esperado:**

``text
¿Todas las raciones cumplen el mínimo de proteína? True
``

**Interpretación:** El resultado es `True` porque cada valor de proteína es igual o superior al mínimo indicado.



## 22. `list()`

**¿Para qué sirve?**  
Crea una lista a partir de un iterable, como un texto o una tupla. La lista es una colección ordenada y modificable.

**Sintaxis general:** ``list(iterable)``

**Ejemplo práctico:**

Convierte los caracteres de una identificación de lote en una lista.



```python
'codigo_lote = "BOV123"
caracteres = list(codigo_lote)

print("Caracteres del código:", caracteres)'

```

**Resultado esperado:**

``text
Caracteres del código: ['B', 'O', 'V', '1', '2', '3']
``

**Interpretación:** Cada carácter del código queda como un elemento independiente de la lista.



## 23. `tuple()`

**¿Para qué sirve?**  
Crea una tupla a partir de un iterable. La tupla conserva el orden de sus elementos y no se puede modificar después de creada.

**Sintaxis general:** ``tuple(iterable)``

**Ejemplo práctico:**

Agrupa las coordenadas de ubicación de un potrero en una tupla.



```python
'coordenadas_potrero = tuple([4.61, -74.08])
print("Coordenadas del potrero:", coordenadas_potrero)'

```

**Resultado esperado:**

``text
Coordenadas del potrero: (4.61, -74.08)
``

**Interpretación:** Las dos coordenadas quedan agrupadas en una colección ordenada e inmutable.



## 24. `set()`

**¿Para qué sirve?**  
Crea un conjunto, que conserva elementos únicos y elimina duplicados. El orden de sus elementos no está garantizado.

**Sintaxis general:** ``set(iterable)``

**Ejemplo práctico:**

Obtiene las identificaciones únicas de los animales observados en varios registros. Se ordenan para mostrar el resultado de forma predecible.



```python
'observaciones = ["Vaca 12", "Vaca 15", "Vaca 12", "Vaca 18"]
animales_unicos = set(observaciones)

print("Pacientes únicos:", sorted(animales_unicos))'

```

**Resultado esperado:**

``text
Identificaciones únicas: ['Vaca 12', 'Vaca 15', 'Vaca 18']
``

**Interpretación:** La identificación repetida se conserva una sola vez dentro del conjunto.



## 25. `dict()`

**¿Para qué sirve?**  
Crea un diccionario que almacena pares de clave y valor. Las claves permiten consultar cada dato por su nombre.

**Sintaxis general:** ``dict(clave=valor)` o `dict(iterable)``

**Ejemplo práctico:**

Guarda algunos datos de identificación y peso de un animal.



```python
'animal = dict(identificacion="Paciente 204", peso_kg=386.5, categoria="Novilla")
print("Animal:", animal["identificacion"])
print("Peso:", animal["peso_kg"], "kg")'

```

**Resultado esperado:**

``text
Animal: Bovino 204\nPeso: 386.5 kg
``

**Interpretación:** El diccionario permite recuperar el valor asociado a cada clave, como la identificación o el peso.



## 26. `format()`

**¿Para qué sirve?**  
Devuelve una representación de un valor con el formato indicado. Por ejemplo, permite mostrar una medición con una cantidad fija de decimales.

**Sintaxis general:** ``format(valor, especificacion)``

**Ejemplo práctico:**

Presenta el consumo diario de alimento con dos decimales.



```python
'consumo_kg = 7.256
consumo_formateado = format(consumo_kg, ".2f")

print("Consumo diario:", consumo_formateado, "kg")'

```

**Resultado esperado:**

``text
Consumo diario: 7.26 kg
``

**Interpretación:** El especificador `.2f` presenta el número con dos cifras decimales.



## 27. `isinstance()`

**¿Para qué sirve?**  
Comprueba si un objeto pertenece a un tipo de dato. Devuelve `True` o `False`.

**Sintaxis general:** ``isinstance(objeto, tipo)``

**Ejemplo práctico:**

Verifica que el peso registrado sea un número decimal (`float`).

**Resultado esperado:**

``text

``

## 29. `dir()`

**¿Para qué sirve?**  
Devuelve una lista de nombres de atributos y métodos disponibles para un objeto.

**Sintaxis general:** ``dir(objeto)``

**Ejemplo práctico:**

Comprueba que una cadena de identificación tenga disponible el método `upper()` para normalizar letras a mayúsculas.



```python
'identificacion = "lote b-12"

print("Identificación normalizada:", identificacion.upper())
print("¿Tiene el método upper?", "upper" in dir(identificacion))'

```

**Resultado esperado:**

``text
Identificación normalizada: LOTE B-12\n¿Tiene el método upper? True
``

**Interpretación:** `dir()` permite revisar los métodos disponibles; `upper()` convierte las letras a mayúsculas.



## 30. `open()`

**¿Para qué sirve?**  
Abre un archivo para leerlo o escribirlo. Se recomienda usar `with` para que el archivo se cierre automáticamente al terminar.

**Sintaxis general:** ``open(ruta, modo, encoding=...)``

**Ejemplo práctico:**

Escribe y luego lee un registro sencillo de pesaje. El archivo se crea en la carpeta de trabajo de Jupyter.



```python
'registro = "Paciente 204: 386.5 kg"

with open("registro_pesaje.txt", "w", encoding="utf-8") as archivo:
    archivo.write(registro)

with open("registro_pesaje.txt", "r", encoding="utf-8") as archivo:
    contenido = archivo.read()

print(contenido)'

```

**Resultado esperado:**

``text
Bovino 204: 386.5 kg
``

**Interpretación:** El modo `w` escribe el texto y el modo `r` lo lee. `with` cierra cada archivo automáticamente.
